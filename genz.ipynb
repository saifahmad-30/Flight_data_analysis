Mental Health analysis

```python
df.head()
```

    Relationship between Daily Usage and Mental Health

```python
df.groupby(pd.cut(df['daily_usage_hours'], bins=[0,2,4,6,10]))['mental_health_score'].mean()
```

```python
plt.scatter(df["daily_usage_hours"], df['mental_health_score'])
plt.xlabel("Daily Usage Hours")
plt.ylabel('Mental Health Score')
plt.title("Usage vs Mental Health")
plt.show()
```

        Platforms associated with Lower Mental Health

```python
df.head()
```
