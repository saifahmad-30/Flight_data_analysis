Bar graphs representing the departure time and arrival time 


```python
df.head()
```

```python
df['departure_time'].value_counts()
```

```python
df['arrival_time'].value_counts()
```

```python
plt.figure(figsize=(16,4))
plt.subplot(1,2,1)

plt.bar(df['departure_time'].value_counts().index, df['departure_time'].value_counts().values , color = {'green','yellow'})

plt.title("Departure time")
plt.xlabel("D- time")
plt.ylabel("frequency")

plt.subplot(1,2,2)
plt.bar(df['arrival_time'].value_counts().index , df['arrival_time'].value_counts().values , color= {'black', 'grey'})
plt.title("Arrival time")
plt.xlabel("A- time")
plt.ylabel("frequency")

plt.show()
```
